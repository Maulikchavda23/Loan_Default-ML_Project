```python
import pandas as pd 
from sklearn.linear_model import LinearRegression 
 
df = pd.read_csv('iphone.csv') 
X = df[['number']]      # feature(s) as a 2-D array -> double brackets 
y = df['Price']         # target as a 1-D array 
 
model = LinearRegression() 
model.fit(X, y)         # <-- this is the 'fit' step students must call

print('Slope (m):', model.coef_) 
print('Intercept (c):', model.intercept_) 
print('Prediction for model 15:', model.predict([[15]]))
```

```python
import numpy as np 
 
df = pd.read_csv('iphone.csv') 
X = df['number'].values.astype(float) 
Y = df['Price'].values.astype(float) 
n = len(X) 
 
m, c = 0.0, 0.0        # 1. start with initial guesses 
lr = 0.001             # 2. learning rate 
epochs = 20000         # 3. number of iterations 
 
loss_history = [] 
for i in range(epochs): 
    Y_pred = m * X + c              # current prediction 
    error  = Y_pred - Y             # predicted - actual 
 
    loss = (1/n) * np.sum(error ** 2)   # MSE loss 
    loss_history.append(loss) 
 
    dm = (2/n) * np.sum(error * X)  # gradient w.r.t. m 
    dc = (2/n) * np.sum(error)      # gradient w.r.t. c 
 
    m = m - lr * dm                 # update m 
    c = c - lr * dc                 # update c 
 
    if i % 4000 == 0: 
        print(f'epoch {i}: loss={loss:.2f}  m={m:.3f}  c={c:.3f}') 
 
print('Final parameters -> m:', m, ' c:', c) 
```

```python
# ============================================================
# TASK 5 - COMPARE ALL 4 CLASSIFICATION MODELS
# Logistic Regression is already completed
# ============================================================

import pandas as pd

from sklearn.ensemble import (
    RandomForestClassifier,
    AdaBoostClassifier,
    GradientBoostingClassifier
)

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)


# ============================================================
# 1. STORE YOUR EXISTING LOGISTIC REGRESSION RESULTS
# ============================================================
# These variables should already exist from your
# Logistic Regression evaluation.

logistic_train_pred = model.predict(X_train)
logistic_test_pred = model.predict(X_test)

logistic_train_accuracy = accuracy_score(
    y_train,
    logistic_train_pred
)

logistic_test_accuracy = accuracy_score(
    y_test,
    logistic_test_pred
)

logistic_precision = precision_score(
    y_test,
    logistic_test_pred,
    zero_division=0
)

logistic_recall = recall_score(
    y_test,
    logistic_test_pred,
    zero_division=0
)

logistic_f1 = f1_score(
    y_test,
    logistic_test_pred,
    zero_division=0
)


# ============================================================
# 2. CREATE THE OTHER THREE MODELS
# ============================================================

models = {

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),

    "AdaBoost": AdaBoostClassifier(
        n_estimators=100,
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=100,
        random_state=42
    )
}


# ============================================================
# 3. STORE LOGISTIC REGRESSION RESULT
# ============================================================

results = []

results.append({
    "Model": "Logistic Regression",
    "Train Accuracy": logistic_train_accuracy,
    "Test Accuracy": logistic_test_accuracy,
    "Precision": logistic_precision,
    "Recall": logistic_recall,
    "F1-score": logistic_f1
})


# ============================================================
# 4. TRAIN AND EVALUATE THE OTHER 3 MODELS
# ============================================================

trained_models = {}

for name, model_new in models.items():

    print("\n" + "=" * 60)
    print("Training:", name)
    print("=" * 60)

    # Train model
    model_new.fit(X_train, y_train)

    # Save trained model
    trained_models[name] = model_new

    # Predictions
    train_pred = model_new.predict(X_train)
    test_pred = model_new.predict(X_test)

    # Train accuracy
    train_accuracy = accuracy_score(
        y_train,
        train_pred
    )

    # Test accuracy
    test_accuracy = accuracy_score(
        y_test,
        test_pred
    )

    # Precision
    precision = precision_score(
        y_test,
        test_pred,
        zero_division=0
    )

    # Recall
    recall = recall_score(
        y_test,
        test_pred,
        zero_division=0
    )

    # F1-score
    f1 = f1_score(
        y_test,
        test_pred,
        zero_division=0
    )

    # Add result to list
    results.append({
        "Model": name,
        "Train Accuracy": train_accuracy,
        "Test Accuracy": test_accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-score": f1
    })


# ============================================================
# 5. CREATE FINAL COMPARISON DATAFRAME
# ============================================================

comparison_df = pd.DataFrame(results)


# ============================================================
# 6. DISPLAY RESULTS
# ============================================================

print("\n")
print("=" * 90)
print("FINAL MODEL COMPARISON")
print("=" * 90)

print(
    comparison_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import (
    RandomForestClassifier,
    AdaBoostClassifier,
    GradientBoostingClassifier
)

import pandas as pd


# ============================================================
# MODELS FOR 5-FOLD CROSS-VALIDATION
# ============================================================

cv_models = {

    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),

    "AdaBoost": AdaBoostClassifier(
        n_estimators=100,
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=100,
        random_state=42
    )
}


# ============================================================
# 5-FOLD CROSS-VALIDATION
# ============================================================

cv_results = []

for name, current_model in cv_models.items():

    print("\n" + "=" * 60)
    print("Cross-validation:", name)
    print("=" * 60)

    scores = cross_val_score(
        current_model,
        X_train,
        y_train,
        cv=5,
        scoring="f1"
    )

    print("Fold scores:", scores)
    print("Mean F1-score:", scores.mean())
    print("Standard deviation:", scores.std())

    cv_results.append({
        "Model": name,
        "CV Mean F1": scores.mean(),
        "CV Std": scores.std()
    })


# ============================================================
# CREATE DATAFRAME
# ============================================================

cv_df = pd.DataFrame(cv_results)

print("\n")
print("=" * 70)
print("5-FOLD CROSS-VALIDATION RESULTS")
print("=" * 70)

print(
    cv_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)
```

```python
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import GridSearchCV

gradient_model = GradientBoostingClassifier(random_state=42)

param_grid = {
    "n_estimators": [50, 100],
    "learning_rate": [0.1],
    "max_depth": [2, 3]
}

grid_search = GridSearchCV(
    gradient_model,
    param_grid,
    scoring="f1",
    cv=3,          # use 3 folds instead of 5 for faster tuning
    n_jobs=-1,
    verbose=1
)

grid_search.fit(X_train, y_train)

print("Best Parameters:", grid_search.best_params_)
print("Best CV F1:", grid_search.best_score_)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

best_model = grid_search.best_estimator_

y_test_pred_tuned = best_model.predict(X_test)

print("Final Test Evaluation")
print("---------------------")
print("Accuracy :", accuracy_score(y_test, y_test_pred_tuned))
print("Precision:", precision_score(y_test, y_test_pred_tuned, zero_division=0))
print("Recall   :", recall_score(y_test, y_test_pred_tuned, zero_division=0))
print("F1-score :", f1_score(y_test, y_test_pred_tuned, zero_division=0))
```
